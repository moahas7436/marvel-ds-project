# Marvel Character Power Analysis

This project explores the characteristics and power levels of Marvel characters using a custom-made dataset.  
The goal is to simulate a real-world team performance analysis using fictional data — showcasing core data science skills such as:

- Data creation using `pandas`
- Aggregation using `.groupby()`
- Sorting and filtering
- Visualization using `matplotlib`
- Clear communication of insights using markdown


## Step 1: Create a Marvel Characters Dataset

We created a small dataset of Marvel characters with fields like Name, Team, Power Level, and Alignment.
This lets us simulate a team performance dataset with fictional characters.


## Insight: Who's the Strongest?

By sorting the dataset by Power Level, we can see which characters are the most powerful. Thor and Hulk top the list, which matches their role in the MCU as heavy-hitters.

```python
import pandas as pd

# Sample Marvel character data
data = {
    'Name': ['Iron Man', 'Captain America', 'Thor', 'Black Widow', 'Hulk', 'Spider-Man', 'Loki'],
    'Team': ['Avengers', 'Avengers', 'Avengers', 'Avengers', 'Avengers', 'Avengers', 'Villains'],
    'Power Level': [85, 80, 95, 70, 90, 78, 88],
    'First Appearance': [1963, 1941, 1962, 1964, 1962, 1962, 1962],
    'Alignment': ['Hero', 'Hero', 'Hero', 'Hero', 'Hero', 'Hero', 'Villain']
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Show the table
df.head()
```

## Step 2: Explore the Marvel Character Data
We'll check basic stats, group by team or alignment, and find the strongest heroes

```python
# See data types and missing values
df.info()

# Get summary of numbers (like Power level)
df.describe()

# How many of each alignment?
df['Alignment'].value_counts()
```

```python
# Sort by Power Level
df.sort_values('Power Level', ascending=False)
```

```python
# Average Power Level per Team
df.groupby('Team')['Power Level'].mean()
```

## Insight: Which Team Is Strongest?

By grouping characters by `Team` and calculating their average power level, we can assess which teams are the most powerful on average.  
In this example, the Avengers dominate — which aligns with their narrative as Earth's mightiest heroes.

## Power Levels Bar Chart

This bar chart displays the power levels of individual Marvel characters.
Thor and Hulk appear as the top-tier heroes, with Black Widow at a lower power level - which matches her role as a non-enhanced human in a superhuman team.

```python
import matplotlib.pyplot as plt

# Bar chart of Power Levels
plt.figure(figsize=(10, 5))
plt.bar(df['Name'], df['Power Level'], color='crimson')
plt.title('Marvel Characters - Power Levels')
plt.xlabel('Character')
plt.ylabel('Power Level (1–100)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
